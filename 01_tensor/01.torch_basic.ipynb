#### PyTorch
: 인공지능과 딥러닝 모델을 학습시키기 위한 파이써 기반의 오픈소스 머신러닝 프레임워크

- 핵심기능
    - Tensor : 데이터를 다차원의 배열로 저장/계산
    - GPU연산 : GPU를 이용한 대량의 계산을 빠르게 처리(병렬처리)
    - 자동 미분(Autograd): 학습에 필요한 기울기를 자동으로 계산
    - 최적화(Optimizer) : 모델의 가중치를 수정하며 학습

#### 초기화 되지 않은 텐서

```python
x = torch.empty(4,2)
x
```

```python
x = torch.rand(4,2)
print(x)
```

#### 0또는 1로 채워진 텐서

```python
x = torch.zeros(4,2)
print(x)
print(x.dtype)

y = torch.ones(2,4)
print(y)
print(y.dtype)
```

#### 사용자가 입력한 값으로 텐서 초기화

```python
a = torch.tensor([1,2,3,4])
print(a)
print(type(a))
print(a.dtype)
print(a.shape)

# 하나라도 실수이면 자동으로 실수타입으로 변환
b= torch.tensor([1,2,3.1,4])
print(b)
print(b.dtype)
```

#### 기초함수

```python
c = torch.tensor([[1,2,3],
                  [4,5,6]])

#행렬이라 각 행 열의 개수가 같아야함. 아래는 오류
# d = torch.tensor([[1,2],[3,4,5]])

print(c)
print(c.shape)
print(c.ndim) # 차원의 수
print(c.numel()) # 전체 데이터의 개수
```

```python
print(torch.zeros(5))
#zeros_like(변수) : 변수에 들어있는 텐서와 동일한 크기를 가지면서 모든 원소가 0인 새로운 텐서 생성
print(torch.zeros_like(c))
print(torch.arange(3,10,2)) # range와 동일
print(torch.arange(0,1,0.1)) #소수점 가능(range에서는 안됨)
print(torch.linspace(0,1,10))#0~1(END 포함) 10개
```

#### 텐서의 연산
:텐서에 대한 수학연산, 삼각함수, 비트연산, 비교연산, 집계 등 제공

```python
d = torch.tensor([1,2,3])
e = torch.tensor([4,5,6])
f = d+e
print(f)
```

```python
g = torch.tensor([[1,2,3],[1,2,3]])
h = torch.tensor([[4,5,6],[1,1,1]])
i = g+h
print(i)

j= g - h
print(j)

print(g*h)
print(g/h)
print(g**2)
```

#### 행렬곱(dot product)

```python
k = torch.tensor([[1,2],[3,4]])
m = torch.tensor([[1,2],[3,4]])

print(k@m)
print(k.matmul(m)) #메소드 방식)
print(torch.mm(k,m)) 
```
