```python
import torch
from torch import nn
```

```python
layer = nn.Conv2d(in_channels=1, out_channels=2, kernel_size=3) #stride =1, padding=0 이 디폴트
layer(torch.randn(32,1,5,5)).shape
```

```python
#in_channels, out_channels, kernel_size는 순서대로 넣으면 생략 가능 | 나머지는 이름 붙여줌(일반적으로)
layer = nn.Conv2d(3,5,3)
print(layer(torch.randn(32,3,5,5)).shape)
print(layer.weight.shape) # [특징(출력), 채널(입력), 행, 열]
```

```python
conv1 = nn.Conv2d(1,8,6,stride=2)
x = torch.randn(32,1,28,28)
print(conv1(x).shape)

conv2 = nn.Conv2d(8,16,3,padding=1)
print(conv2(conv1(x)).shape)

maxpool = nn.MaxPool2d(kernel_size=2,stride=2)
print(maxpool(conv2(conv1(x))).shape)
```

```python
#보통 kernel_size와 stride 두 값을 똑같이 넣기 때문에 하나만 넣어줘도됨
maxpool = nn.MaxPool2d(2)
x = torch.randn(1,6,6)
print(x)
print(maxpool(x))
print(maxpool(torch.randn(32,3,6,6)).shape)
```

```python
maxpool = nn.AvgPool2d(2)
x = torch.randn(1,6,6)
print(x)
print(maxpool(x))
```

```python
class CNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.conv1 = nn.Conv2d(1,8,6,stride=2)
        self.conv2 = nn.Conv2d(8,16,3,padding=1)
        self.maxpool = nn.MaxPool2d(2)
        self.fc = nn.Linear(16*6*6, 10)

    def forward(self, x):
        x = self.conv1(x)
        print('conv1 :', x.shape)
        x = self.conv2(x)
        print('conv2 :', x.shape)
        x = maxpool(x)
        print('maxpool :', x.shape)
        x = torch.flatten(x, start_dim=1)
        print('flatten :', x.shape)
        x = self.fc(x)
        return x

x = torch.randn(32,1,28,28)
model = CNN()
print(model(x).shape)

# (input + 2*padding - kernel)/stride <== 계산 수식
```
