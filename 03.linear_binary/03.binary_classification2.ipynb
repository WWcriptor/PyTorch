#### 비선형 이진분류(동심원 데이터)

```python
import torch
import matplotlib.pyplot as plt
```

#### step 1. 데이터 생성

```python
N = 100

radius1 = 0.3 + 0.05 * torch.randn(int(N/2),1) # 0.3 주변 값(50개)
radius2 = 1 + 0.01 * torch.randn(int(N/2),1) # 1 주변 값(50개)

# 각도 만들기
theta1 = 2*torch.pi*torch.rand(int(N/2),1) - torch.pi # -파이 ~ 파이 사이의 값
theta2 = 2*torch.pi*torch.rand(int(N/2),1) - torch.pi

# (x,y)좌표로 바꾸기. 극좌표를 직교좌표로 변환
class1_data = torch.hstack([radius1*torch.cos(theta1), radius1*torch.sin(theta1)])
class2_data = torch.hstack([radius2*torch.cos(theta2), radius2*torch.sin(theta2)])

class1_label = torch.ones(int(N/2),1)
class2_label = torch.zeros(int(N/2),1)

X = torch.vstack([class1_data, class2_data])
y = torch.vstack([class1_label, class2_label])

```

```python
plt.plot(class1_data[:,0], class1_data[:,1], 'o')
plt.plot(class2_data[:,0], class2_data[:,1], 'rs')
plt.xlabel('x1')
plt.ylabel('x2')
plt.grid()
```

#### step 2. 모델 만들기

```python
from torch import nn

class MLP(nn.Module):
    def __init__(self):
        super().__init__()

        self.linear = nn.Sequential(nn.Linear(2,100),
                                    nn.Sigmoid(), 
                                    nn.Linear(100,1), 
                                    nn.Sigmoid())

        # 비선형에서 너무 간단한 시퀀셜은 학습이 잘 되지 않음
        # self.linear = nn.Sequential(nn.Linear(2,1), 
        #                             nn.Sigmoid()) 

        # self.linear = nn.Sequential(nn.Linear(2,100),
        #                                     nn.Sigmoid(), 
        #                                     nn.Linear(100,100),
        #                                     nn.Sigmoid(), 
        #                                     *[nn.Sequential(nn.Linear(100,100),
        #                                     nn.Sigmoid()) for _ in range(10) ],
        #                                     nn.Linear(100,1), 
        #                                     nn.Sigmoid())

        self.linear = nn.Sequential(nn.Linear(2,100),
                                            nn.ReLU(), 
                                            *[nn.Sequential(nn.Linear(100,100),
                                            nn.ReLU()) for _ in range(10) ],
                                            nn.Linear(100,1), 
                                            nn.Sigmoid())

    def forward(self, x):
        x = self.linear(x)
        return x
```

```python
model = MLP()
```

#### step 3. 모델 학습시키기

```python
from torch import optim

LR = 1e-4
EPOCH = 200

optimizer = optim.Adam(model.parameters(), lr=LR)
# optimizer = optim.SGD(model.parameters(), lr=LR) # 선형이 아니라 잘 안 됨
cirterion = nn.BCELoss()

loss_history= []

model.train()
for ep in range(EPOCH):
    y_hat = model(X)

    loss = cirterion(y_hat, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    loss_history += [loss.item()]

    print(f'Epoch : {ep+1},  Train Loss : {loss.item():.4f}')
    print('-'*20)
```

```python
plt.plot(range(1, EPOCH+1), loss_history)
plt.xlabel('Epoch')
plt.ylabel('Loss')
```

#### step 4. 모델 테스트

```python
x1_test = torch.linspace(-2,2,30)
x2_test = torch.linspace(-2,2,30)

X1_test, X2_test = torch.meshgrid(x1_test, x2_test)
X_test = torch.cat([X1_test.unsqueeze(dim=2), X2_test.unsqueeze(dim=2)],dim=2)

model.eval() # 테스트 모드로 전환
with torch.no_grad():
    y_hat = model(X_test)

Y_hat = y_hat.squeeze()

plt.figure(figsize=(10,9))
ax = plt.axes(projection='3d')
ax.view_init(elev=25, azim=-140)

ax.plot_surface(X1_test, X2_test, Y_hat.numpy(), cmap='bone', alpha=0.2)
plt.plot(class1_data[:,0],class1_data[:,1],class1_label, 'bo')
plt.plot(class2_data[:,0],class2_data[:,1],class2_label, 'ro')
```
