#### 문제 02

실험 장비에서 측정한 양수 데이터가 다음과 같음.

```python

```

**요구사항**
1. 각 원소의 제곱근을 `sqrt_data`에 저장.
2. 각 원소의 제곱을 `square_data`에 저장.
3. 각 원소에 자연상수 \(e\)를 밑으로 하는 지수함수를 적용.


1. 전체 생산량의 합계와 평균 계산.

````python
import pandas as pd
from torch.utils.data import DataLoader, TensorDataset

df = pd.DataFrame({'학습시간' : [12,4,5,3,7,3,15,7],
                   '문제풀이 횟수' : [15,23,3,6,14,25,45,2],
                   '점수' : [90,80,85,78,98,79,97,86]})

x = df.drop('점수',axis = 1)
y = df.loc[:, '점수']

x_np = x.values
y_np = y.values

y_np = y_np.reshape(-1,1)

x_tensor = torch.tensor(x_np, dtype=torch.float32)
y_tensor = torch.tensor(y_np, dtype=torch.float32)

dataset = TensorDataset(x_tensor, y_tensor)

print(len(dataset))
print(dataset[:3])

data_loader = DataLoader(dataset, batch_size=3, shuffle=False)

print(len(data_loader))
for x_batch, y_batch in data_loader:
    print(x_batch)
    print(y_batch)

data_loader2 = DataLoader(dataset, batch_size=3, shuffle=True)
for x_batch, y_batch in data_loader2:
    print(x_batch)
    print(y_batch)
    break
````
